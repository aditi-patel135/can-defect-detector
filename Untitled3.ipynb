```python
df1=df.copy()
df.head(10)
```

```python
df1["Age"].isna().sum()
print("AGE", df1.Age)
df["Age"].mean()
```

```python
df1["Age"] = df1["Age"].fillna(df1["Age"].mean())
print("missing age before",df1["Age"].isna().sum())
df1.head(10)
df1["Age"].head(10)
```

```python
num_imputer = SimpleImputer(strategy='median')
df1["Age"]=num_imputer.fit_transform(df1[["Age"]])
num_imputer.fit(df1[["Age"]])
print(num_imputer.statistics_)
```

```python
num_imputer = SimpleImputer(strategy='most_frequent')
df1["Embarked"]=num_imputer.fit_transform(df1[["Embarked"]])
num_imputer.fit(df1[["Embarked"]])
print(num_imputer.statistics_)
```
